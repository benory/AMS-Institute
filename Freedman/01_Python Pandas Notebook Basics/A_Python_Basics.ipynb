## 4. Working with Text (Strings)

Text in Python is called a **string**. Strings are written inside quotes — either `'single'` or `"double"`.

You can:
- **Concatenate** (join) strings with `+`
- **Embed variables** inside strings using an f-string: `f"Hello, {name}!"`

Run the cell below to try these out.

```python
first_name = "Guillaume"
last_name  = "Dufay"

# Concatenation
full_name = first_name + " " + last_name
print("Full name:", full_name)


# f-string formatting
year = 1400
print(f"{full_name} was born in about {year}.")
```

## 5. String Methods

Strings come with built-in **methods** you call with dot notation: `my_string.method()`

| Method | What it does |
|--------|--------------|
| `.upper()` / `.lower()` | Change case |
| `.strip()` | Remove leading/trailing whitespace |
| `.split(sep)` | Split into a list at each separator |
| `.replace(old, new)` | Swap one substring for another |
| `.find(sub)` | Return the index of the first match (or -1) |
| `.startswith(s)` / `.endswith(s)` | Test how a string begins or ends |
| `len(s)` | Count characters (a built-in function, not a method) |

```python
title = " L'incoronazione di Popeye "

# Strip whitespace, then change case
print(title.strip())
print(title.strip().upper())
print(title.strip().lower())

# Replace a substring
corrected = title.replace("Popeye", "Poppea")
print(corrected)

# Split a full name into a list of parts
composer = "Claudio Monteverdi"
name_parts = composer.split(" ")
print(name_parts)

# now revserse the list and join it back together
reversed_name = ", ".join(reversed(name_parts))
print(reversed_name)    

# startswith / endswith
genre = "opera"
print(genre.startswith("sing"))   # True
print(genre.endswith("opera"))    # False

# Length and find
composer = "Claudio Monteverdi"
print("Length:", len(composer))
print("'van' starts at index:", composer.find("van"))
```

## 6 Convert Between Types 

Need to add a number to a string? Or extract a number from a string? Use these functions to convert between types!

```python
# this is an int, not a str!
birth_year = 1567 

# we need to convert the int birth_year to a str before concatenating it with other strings
year_str = str(birth_year)
print("Born in " + year_str)       # str concatenation now works

# on the other hand, if the date is stored as a str, we can't do arithmetic on it until we convert it to an int
birth_year_str = "1567" # this is a str, not an int!
print("Born in " + birth_year_str)  # str concatenation works!

# but we can't do arithmetic on a str, so this will give an error:
# print(birth_year_str + 10)        # this will give an error
year_int = int("1567")
print(year_int + 10)               # arithmetic on the converted int
```

## 7. Working with Lists

A **list** is an ordered, changeable collection written in square brackets `[]`.  
Items can be any type; duplicates are allowed.

| Operation | Code |
|-----------|------|
| Access by index | `my_list[0]` (first) &nbsp;&nbsp; `my_list[-1]` (last) |
| Add an item | `my_list.append(item)` |
| Remove an item | `my_list.remove(item)` |
| Count items | `len(my_list)` |
| Unique items | `len(set(my_list))` |
| Sort | `my_list.sort()` or `my_list.sort(reverse=True)` |
| Compact filter | `[x for x in my_list if condition]` |

```python
genres = ["madrigal", "chanson", "motet", "fantasia", "mass"]

# Access by index
print("First:", genres[0])
print("Last: ", genres[-1])

# Length and unique count
print("Total:", len(genres))
print("Unique:", len(set(genres)))

# Add and remove
genres.append("organum")
genres.remove("fantasia")
print("After edits:", genres)

# Sort alphabetically
genres.sort()
print("Sorted:", genres)

# List comprehension: keep only genres starting with "m"
m_genres = [g for g in genres if g.startswith("m")]
print("M-genres:", m_genres)
```

## 8. Working with Dictionaries

A **dictionary** stores data as **key-value pairs** inside curly braces `{}`.  
Keys are usually strings; values can be any type.

| Operation | Code |
|-----------|------|
| Access a value | `my_dict["key"]` |
| List all keys | `my_dict.keys()` |
| List all values | `my_dict.values()` |
| Add or update | `my_dict["key"] = value` |
| Remove a key | `my_dict.pop("key")` |
| Loop over pairs | `for k, v in my_dict.items():` |

A **list of dictionaries** is one of the most common data structures — think of each dict as one row in a spreadsheet.

```python
# A single dictionary describing one musical work
work = {
    "composer_first_name": "Claudio",
    "composer_last_name":  "Monteverdi",
    "title":               "L'incoronazione di Popeye",
    "genre":               "opera",
    "date":                "1640",
    "premiere_city":       "Venice"
}

# Access a value by key
print(work["title"])

# Add a new key
work["librettist"] = "Giovanni FrancescoBusenello"
print(list(work.keys()))

# Loop over all key-value pairs
for key, value in work.items():
    print(f"  {key}: {value}")

# -----------------------------------------------
# A list of dictionaries (like rows in a table)
operas = [
    {"composer": "Monteverdi",  "title": "L'incoronazione di Popeye", "year": 1640},
    {"composer": "Monteverdi",   "title": "L'orfeo", "year": 1605},
    {"composer": "Handel", "title": "Orlando", "year": 1733},
]

# Print every title
for opera in operas:
    print(opera["title"])

# Filter: only operas premiered after 1700
recent = [o for o in operas if o["year"] > 1700]
print("\nAfter 1700:")
for o in recent:
    print(f'  {o["composer"]} — {o["title"]} ({o["year"]})')
```
