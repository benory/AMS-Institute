## 3. Basic Arithmetic

Python can be used as a calculator. Here are the basic arithmetic operators:

| Symbol | Operation      |
|--------|----------------|
| `+`    | Addition       |
| `-`    | Subtraction    |
| `*`    | Multiplication |
| `/`    | Division       |
| `**`   | Exponentiation |

Run the cell below to see some examples.

```python
a = 12
b = 5

print("Addition:      ", a + b)
print("Subtraction:   ", a - b)
print("Multiplication:", a * b)
print("Division:      ", a / b)
print("Exponentiation:", a ** b)
```

## 4. Working with Text (Strings)

Text in Python is called a **string**. Strings are written inside quotes — either `'single'` or `"double"`.

You can:
- **Concatenate** (join) strings with `+`
- **Repeat** a string with `*`
- **Embed variables** inside strings using an f-string: `f"Hello, {name}!"`

Run the cell below to try these out.

```python
first_name = "Wolfgang Amadeus"
last_name  = "Mozart"

# Concatenation
full_name = first_name + " " + last_name
print("Full name:", full_name)

# Repetition
print("Echo: " + (first_name + "! ") * 3)

# f-string formatting
year = 1756
print(f"{full_name} was born in {year}.")
```
